```python
import pandas as pd

train = pd.read_csv(
    "../dataset/train.txt",
    sep="\t",
    header=None,
    names=["head", "relation", "tail"]
)

train.head()
```

```python
print("Number of triples:", len(train))

entities = set(train["head"]) | set(train["tail"])

print("Number of unique entities:", len(entities))

print("Number of unique relations:", train["relation"].nunique())
```

```python
train["relation"].value_counts()
```

```python
profession_triples = train[
    train["relation"] == "/people/person/profession"
]

profession_triples.head(10)
```

```python
film_triples = train[
    train["relation"] == "/film/actor/film./film/performance/film"
]

film_triples.head(10)
```

```python
print("Profession triples:", len(profession_triples))
print("Film triples:", len(film_triples))
```

```python
print("Unique heads:", profession_triples["head"].nunique())
print("Unique tails:", profession_triples["tail"].nunique())
```

```python
profession_triples["tail"].value_counts().head(10)
```

```python
print("Total profession triples:", len(profession_triples))
print("Unique people/entities:", profession_triples["head"].nunique())
print("Unique professions/entities:", profession_triples["tail"].nunique())
```
